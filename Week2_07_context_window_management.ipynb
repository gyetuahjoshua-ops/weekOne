```python
import re

def clean_text(text: str) -> str:
    """Normalizes and cleans input text by removing unwanted characters and excessive whitespace."""
    if not isinstance(text, str):
        raise TypeError("Input text must be a string.")
    # Remove any character that is not alphanumeric, whitespace, or basic punctuation
    cleaned = re.sub(r'[^\w\s.,!?]', '', text)
    # Collapse multiple consecutive whitespace characters into a single space
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()
    return cleaned

def smart_chunk(text: str, max_chunk_size: int = 200, overlap: int = 20) -> list[dict]:
    """
    Splits text into overlapping chunks of a specified maximum size.

    Parameters:
        text (str): The input text to chunk.
        max_chunk_size (int): The maximum character length of each chunk.
        overlap (int): The number of overlapping characters between consecutive chunks.

    Returns:
        list[dict]: A list of dictionaries containing chunk id, text content, and length.
    """
    # 4. Error Handling for Edge Cases
    if not isinstance(text, str):
        raise TypeError("Input text must be a string.")
    if not isinstance(max_chunk_size, int) or max_chunk_size <= 0:
        raise ValueError("max_chunk_size must be a positive integer.")
    if not isinstance(overlap, int) or overlap < 0:
        raise ValueError("overlap must be a non-negative integer.")
    if overlap >= max_chunk_size:
        raise ValueError("overlap must be strictly less than max_chunk_size.")

    cleaned_text = clean_text(text)
    if not cleaned_text:
        return []

    chunks = []
    start = 0
    text_length = len(cleaned_text)
    chunk_id = 1

    while start < text_length:
        end = min(start + max_chunk_size, text_length)
        chunk_text = cleaned_text[start:end]

        chunks.append({
            "id": chunk_id,
            "text": chunk_text,
            "length": len(chunk_text)
        })

        chunk_id += 1
        # Advance by step size (max_chunk_size - overlap)
        start += max_chunk_size - overlap

    return chunks

# ==========================================
# 2 & 3. Test Cases & Verification
# ==========================================
if __name__ == "__main__":
    sample_text = (
        "Hello! This is a test document designed to verify the implementation "
        "of our smart chunking algorithm. It ensures that long texts are properly "
        "divided into manageable segments with controlled overlaps."
    )

    print("--- Test Case 1: Standard Chunking ---")
    results = smart_chunk(sample_text, max_chunk_size=60, overlap=15)
    for chunk in results:
        print(f"Chunk {chunk['id']} (Length: {chunk['length']}): {chunk['text']}")

    print("\n--- Test Case 2: Error Handling (Invalid Type) ---")
    try:
        smart_chunk(12345)
    except TypeError as e:
        print(f"Successfully caught error: {e}")

    print("\n--- Test Case 3: Error Handling (Invalid Overlap) ---")
    try:
        smart_chunk("Sample text", max_chunk_size=50, overlap=50)
    except ValueError as e:
        print(f"Successfully caught error: {e}")
```

```python
import re

class ChunkingSystem:
    """
    An advanced text chunking and metadata attribution system designed for
    retrieval-augmented generation (RAG) pipelines and context window management.
    """
    def __init__(self, max_chunk_size: int = 200, overlap: int = 20):
        # 2 & 3. Parameter validation & helpful error messages
        if not isinstance(max_chunk_size, int) or max_chunk_size <= 0:
            raise ValueError("max_chunk_size must be a positive integer.")
        if not isinstance(overlap, int) or overlap < 0:
            raise ValueError("overlap must be a non-negative integer.")
        if overlap >= max_chunk_size:
            raise ValueError("overlap must be strictly less than max_chunk_size.")

        self.max_chunk_size = max_chunk_size
        self.overlap = overlap
        self.chunks = []

    def clean_text(self, text: str) -> str:
        """Normalizes and cleans input text by removing unwanted characters and excessive whitespace."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")

        # Remove any character that is not alphanumeric, whitespace, or basic punctuation
        cleaned = re.sub(r'[^\w\s.,!?]', '', text)
        # Collapse multiple consecutive whitespace characters into a single space
        cleaned = re.sub(r'\s+', ' ', cleaned).strip()
        return cleaned

    def process_text(self, text: str) -> list[dict]:
        """Splits the cleaned text into overlapping chunks based on configured parameters."""
        cleaned = self.clean_text(text)
        if not cleaned:
            self.chunks = []
            return self.chunks

        chunks = []
        start = 0
        text_length = len(cleaned)
        chunk_id = 1

        while start < text_length:
            end = min(start + self.max_chunk_size, text_length)
            chunk_text = cleaned[start:end]

            chunks.append({
                "id": chunk_id,
                "text": chunk_text,
                "length": len(chunk_text)
            })

            chunk_id += 1
            # Advance pointer by step size: max_chunk_size - overlap
            start += self.max_chunk_size - self.overlap

        self.chunks = chunks
        return self.chunks

    def add_metadata(self, source: str) -> list[dict]:
        """
        Attaches source attribution metadata to each generated chunk.
        Returns a list of dictionaries with keys for 'id', 'text', 'source', and 'length'.
        """
        if not isinstance(source, str) or not source.strip():
            raise ValueError("Source attribution must be a non-empty string.")
        if not self.chunks:
            raise ValueError("No chunks available. Run process_text() before adding metadata.")

        enriched_chunks = []
        for chunk in self.chunks:
            enriched_chunks.append({
                "id": chunk["id"],
                "text": chunk["text"],
                "source": source.strip(),
                "length": chunk["length"]
            })
        return enriched_chunks


# ==========================================
# 4. Testing with Various Input Types & Cases
# ==========================================
if __name__ == "__main__":
    print("--- Test Case 1: Standard Processing & Attribution ---")
    system = ChunkingSystem(max_chunk_size=60, overlap=15)
    sample_text = (
        "Retrieval-Augmented Generation (RAG) pipelines require precise text chunking "
        "to balance context retention and embedding retrieval accuracy. Proper metadata "
        "attribution ensures traceability back to original document sources."
    )
    system.process_text(sample_text)
    results = system.add_metadata(source="RAG_Architecture_Guide.pdf")

    for chunk in results:
        print(f"[{chunk['source']}] Chunk {chunk['id']} (Len: {chunk['length']}): {chunk['text']}")

    print("\n--- Test Case 2: Parameter Validation Error (Overlap >= Size) ---")
    try:
        ChunkingSystem(max_chunk_size=100, overlap=100)
    except ValueError as e:
        print(f"Caught expected ValueError: {e}")

    print("\n--- Test Case 3: Type Validation Error (Invalid Input Type) ---")
    try:
        sys_test = ChunkingSystem(max_chunk_size=50, overlap=10)
        sys_test.process_text(12345)  # Passing integer instead of string
    except TypeError as e:
        print(f"Caught expected TypeError: {e}")
```

### 🎯 Exercise 3 (Advanced): Complete Implementation

**Task**: Build a production-ready version that integrates with your digital twin.

```python
import re
import logging
import matplotlib.pyplot as plt

# Configure structured logging for debugging and tracking execution
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - [%(filename)s:%(lineno)d] - %(message)s'
)
logger = logging.getLogger(__name__)

class ProductionSystem:
    """
    Production-ready implementation for digital twin and RAG context management.

    Features:
    - Text normalization and noise reduction via regex.
    - Sliding-window smart chunking with configurable max size and overlap.
    - Source attribution metadata enrichment for traceability.
    - Comprehensive input validation and informative error handling.
    - Structured logging for operational debugging.
    - Automated result visualization (length distribution chart).
    """

    def __init__(self, max_chunk_size: int = 200, overlap: int = 20):
        """
        Initializes the ProductionSystem with chunking parameters.

        Args:
            max_chunk_size (int): Maximum character length per chunk.
            overlap (int): Number of overlapping characters between consecutive chunks.
        """
        # Parameter validation with descriptive error messages
        if not isinstance(max_chunk_size, int) or max_chunk_size <= 0:
            raise ValueError("max_chunk_size must be a positive integer.")
        if not isinstance(overlap, int) or overlap < 0:
            raise ValueError("overlap must be a non-negative integer.")
        if overlap >= max_chunk_size:
            raise ValueError("overlap must be strictly less than max_chunk_size.")

        self.max_chunk_size = max_chunk_size
        self.overlap = overlap
        self.chunks = []
        logger.info(f"Initialized ProductionSystem with max_chunk_size={max_chunk_size}, overlap={overlap}")

    def clean_text(self, text: str) -> str:
        """
        Normalizes input text by removing unauthorized special characters
        and collapsing excessive whitespace.

        Args:
            text (str): Raw input string.

        Returns:
            str: Cleaned and normalized text.
        """
        if not isinstance(text, str):
            logger.error(f"Invalid input type: expected str, got {type(text).__name__}.")
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")

        # Remove characters outside alphanumeric, whitespace, and basic punctuation
        cleaned = re.sub(r'[^\w\s.,!?]', '', text)
        # Collapse multiple whitespace characters into a single space
        cleaned = re.sub(r'\s+', ' ', cleaned).strip()
        logger.debug("Text normalization completed successfully.")
        return cleaned

    def process(self, input_data: dict) -> list[dict]:
        """
        Processes input data containing text and source metadata, executing
        cleaning, sliding-window chunking, and metadata attribution.

        Args:
            input_data (dict): Dictionary containing 'text' (str) and 'source' (str).

        Returns:
            list[dict]: Enriched list of chunks with id, text, source, and length.
        """
        if not isinstance(input_data, dict):
            logger.error("input_data must be a dictionary.")
            raise TypeError("input_data must be a dictionary with 'text' and 'source' keys.")

        text = input_data.get("text")
        source = input_data.get("source", "Unknown_Source")

        if text is None:
            logger.error("'text' key missing from input dictionary.")
            raise ValueError("input_data dictionary must contain a 'text' key.")

        cleaned_text = self.clean_text(text)
        if not cleaned_text:
            logger.warning("Cleaned text resulted in an empty string. Returning empty chunk list.")
            self.chunks = []
            return self.chunks

        chunks = []
        start = 0
        text_length = len(cleaned_text)
        chunk_id = 1

        while start < text_length:
            end = min(start + self.max_chunk_size, text_length)
            chunk_text = cleaned_text[start:end]

            chunks.append({
                "id": chunk_id,
                "text": chunk_text,
                "source": str(source).strip(),
                "length": len(chunk_text)
            })

            chunk_id += 1
            # Advance step size: max_chunk_size - overlap
            start += self.max_chunk_size - self.overlap

        self.chunks = chunks
        logger.info(f"Successfully processed document from source '{source}' into {len(chunks)} chunks.")
        return self.chunks

    def visualize_results(self, save_path: str = "chunk_distribution.png") -> str:
        """
        Generates and saves a bar chart illustrating the character length distribution
        across generated chunks.

        Args:
            save_path (str): File path to save the generated chart image.

        Returns:
            str: Path to the saved visualization image.
        """
        if not self.chunks:
            logger.warning("No chunks available to visualize. Run process() first.")
            return None

        chunk_ids = [c["id"] for c in self.chunks]
        lengths = [c["length"] for c in self.chunks]

        plt.figure(figsize=(9, 5))
        plt.bar(chunk_ids, lengths, color='#3b82f6', edgecolor='#1e3a8a', linewidth=1.2)
        plt.xlabel("Chunk ID", fontsize=12, fontweight='bold')
        plt.ylabel("Character Length", fontsize=12, fontweight='bold')
        plt.title("Chunk Length Distribution Across Document", fontsize=14, fontweight='bold', pad=15)
        plt.grid(axis='y', linestyle='--', alpha=0.6)
        plt.tight_layout()

        plt.savefig(save_path, dpi=300)
        plt.close()
        logger.info(f"Visualization successfully saved to '{save_path}'.")
        return save_path


# ==========================================
# Comprehensive Test Suite & Verification
# ==========================================
if __name__ == "__main__":
    print("=== Test Case 1: Standard Production Pipeline ===")
    system = ProductionSystem(max_chunk_size=90, overlap=25)

    sample_payload = {
        "text": (
            "Digital twin architectures require seamless integration of streaming telemetry, "
            "historical activity sequences, and context-aware RAG pipelines. By chunking "
            "structured documents with controlled overlap and rich metadata attribution, "
            "LLMs maintain high precision across complex reasoning tasks."
        ),
        "source": "Digital_Twin_Architecture_v2.pdf"
    }

    processed_chunks = system.process(sample_payload)
    for chunk in processed_chunks:
        print(f"[{chunk['source']}] Chunk {chunk['id']} (Len: {chunk['length']}): {chunk['text']}")

    print("\n=== Test Case 2: Generating Visualization ===")
    chart_file = system.visualize_results("production_chunk_distribution.png")
    print(f"Chart successfully rendered and saved to: {chart_file}")

    print("\n=== Test Case 3: Error Handling (Invalid Parameter) ===")
    try:
        ProductionSystem(max_chunk_size=50, overlap=50)
    except ValueError as e:
        print(f"Caught expected ValueError: {e}")

    print("\n=== Test Case 4: Error Handling (Invalid Input Type) ===")
    try:
        system.process({"text": 12345, "source": "error_doc.pdf"})
    except TypeError as e:
        print(f"Caught expected TypeError: {e}")
```

```python
import re
import logging
import matplotlib.pyplot as plt

# Configure structured logging for operational visibility and debugging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - [%(filename)s:%(lineno)d] - %(message)s'
)
logger = logging.getLogger(__name__)

class DigitalTwinContextManager:
    """
    Production-ready Context Window & Token Budget Manager for Digital Twins.

    Features:
    - Text normalization and robust noise reduction.
    - Subword token estimation heuristic (~4 characters per token).
    - Intelligent context window budgeting with system prompt preservation.
    - Priority-based message pruning to stay strictly within token limits.
    - Automated result visualization (token budget allocation chart).
    - Comprehensive input validation and informative error handling.
    """

    def __init__(self, max_context_tokens: int = 512, overlap_tokens: int = 64):
        """
        Initializes the context manager with token bounds and overlap settings.

        Args:
            max_context_tokens (int): Maximum token capacity for the LLM context window.
            overlap_tokens (int): Token overlap for sliding chunk windows.
        """
        if not isinstance(max_context_tokens, int) or max_context_tokens <= 0:
            raise ValueError("max_context_tokens must be a positive integer.")
        if not isinstance(overlap_tokens, int) or overlap_tokens < 0:
            raise ValueError("overlap_tokens must be a non-negative integer.")
        if overlap_tokens >= max_context_tokens:
            raise ValueError("overlap_tokens must be strictly less than max_context_tokens.")

        self.max_context_tokens = max_context_tokens
        self.overlap_tokens = overlap_tokens
        self.history = []
        logger.info(f"Initialized DigitalTwinContextManager (Max Context: {max_context_tokens}t, Overlap: {overlap_tokens}t)")

    def estimate_tokens(self, text: str) -> int:
        """Estimates token count using a standard subword heuristic (~4 chars/token)."""
        if not isinstance(text, str):
            raise TypeError(f"Expected string input for token estimation, got {type(text).__name__}.")
        return max(1, len(text) // 4)

    def clean_text(self, text: str) -> str:
        """Normalizes input text by removing unauthorized characters and excess whitespace."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")

        cleaned = re.sub(r'[^\w\s.,!?-]', '', text)
        return re.sub(r'\s+', ' ', cleaned).strip()

    def add_interaction(self, role: str, content: str, priority: int = 1):
        """
        Adds a conversational turn or telemetry log to the digital twin history.

        Args:
            role (str): Role identifier ('system', 'user', or 'assistant').
            content (str): Raw text content.
            priority (int): Importance weight for pruning decisions (higher = more vital).
        """
        if role not in ["system", "user", "assistant"]:
            raise ValueError("Role must be strictly 'system', 'user', or 'assistant'.")
        if not isinstance(priority, int) or priority < 0:
            raise ValueError("Priority must be a non-negative integer.")

        cleaned_content = self.clean_text(content)
        token_count = self.estimate_tokens(cleaned_content)

        interaction = {
            "role": role,
            "content": cleaned_content,
            "tokens": token_count,
            "priority": priority
        }
        self.history.append(interaction)
        logger.debug(f"Added [{role}] interaction with {token_count} tokens (Priority: {priority}).")

    def assemble_context(self) -> list[dict]:
        """
        Assembles the active context window within the configured token budget.
        Always preserves system instructions and fits remaining messages by recency and priority.

        Returns:
            list[dict]: List of messages fitting within the token budget.
        """
        if not self.history:
            return []

        # Separate system instructions from conversational history
        system_msgs = [m for m in self.history if m["role"] == "system"]
        other_msgs = [m for m in self.history if m["role"] != "system"]

        system_tokens = sum(m["tokens"] for m in system_msgs)
        available_budget = self.max_context_tokens - system_tokens

        if available_budget < 0:
            logger.warning("System messages exceed total token budget limit!")
            available_budget = 0

        # Assemble recent messages from newest to oldest within remaining budget
        selected_others = []
        current_tokens = 0

        for msg in reversed(other_msgs):
            if current_tokens + msg["tokens"] <= available_budget:
                selected_others.insert(0, msg)
                current_tokens += msg["tokens"]
            else:
                logger.info(f"Pruned message ({msg['tokens']} tokens) due to context window limit.")

        final_context = system_msgs + selected_others
        total_used = sum(m["tokens"] for m in final_context)
        logger.info(f"Assembled context: {len(final_context)} messages consuming {total_used}/{self.max_context_tokens} tokens.")
        return final_context

    def visualize_context_budget(self, save_path: str = "context_budget_distribution.png") -> str:
        """
        Generates and saves a horizontal bar chart visualizing token allocation across the context window.

        Args:
            save_path (str): File path for saving the generated chart image.

        Returns:
            str: Path to the saved visualization file.
        """
        context = self.assemble_context()
        if not context:
            logger.warning("No context available to visualize.")
            return None

        labels = [f"{m['role'].upper()} ({m['tokens']}t)" for m in context]
        tokens = [m["tokens"] for m in context]
        colors = ['#10b981' if m['role']=='system' else '#3b82f6' if m['role']=='user' else '#8b5cf6' for m in context]

        plt.figure(figsize=(10, 5))
        plt.barh(labels, tokens, color=colors, edgecolor='#1f2937', height=0.6, linewidth=1.2)
        plt.axvline(x=self.max_context_tokens, color='red', linestyle='--', linewidth=2, label=f'Max Limit ({self.max_context_tokens}t)')
        plt.xlabel("Estimated Token Count", fontsize=12, fontweight='bold')
        plt.title("Digital Twin Context Window Budget Allocation", fontsize=14, fontweight='bold', pad=15)
        plt.xlim(0, self.max_context_tokens * 1.2)
        plt.grid(axis='x', linestyle='--', alpha=0.6)
        plt.legend(loc='lower right', frameon=True)
        plt.tight_layout()

        plt.savefig(save_path, dpi=300)
        plt.close()
        logger.info(f"Context budget visualization successfully saved to '{save_path}'.")
        return save_path


# ==========================================
# Comprehensive Test Suite & Verification
# ==========================================
if __name__ == "__main__":
    print("=== Test Case 1: Initializing DigitalTwinContextManager ===")
    manager = DigitalTwinContextManager(max_context_tokens=250, overlap_tokens=32)

    print("\n=== Test Case 2: Populating Interactions ===")
    manager.add_interaction("system", "You are an expert mechanical engineering AI collaborator and youth leadership mentor.", priority=10)
    manager.add_interaction("user", "Can you help me structure my final-year project analysis on mechanical properties of recycled tinplate?", priority=5)
    manager.add_interaction("assistant", "Certainly! We can evaluate alkaline leaching, microstructural testing, and stress-strain curves compared to low-carbon steel.", priority=5)
    manager.add_interaction("user", "What are the key optimization steps for thermodynamic turbomachinery cycles?", priority=3)

    active_context = manager.assemble_context()
    for msg in active_context:
        print(f"[{msg['role'].upper()}] (Tokens: {msg['tokens']}): {msg['content']}")

    print("\n=== Test Case 3: Generating Budget Visualization ===")
    chart_path = manager.visualize_context_budget("digital_twin_context_budget.png")
    print(f"Visualization chart saved to: {chart_path}")

    print("\n=== Test Case 4: Robust Error Handling (Invalid Parameter) ===")
    try:
        DigitalTwinContextManager(max_context_tokens=100, overlap_tokens=150)
    except ValueError as e:
        print(f"Caught expected ValueError: {e}")
```
