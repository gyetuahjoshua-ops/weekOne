```python
import re

def estimate_and_clean_text(text: str) -> dict:
    """
    Normalizes input text and estimates its token count using the standard
    heuristic of ~4 characters per token.

    Parameters:
        text (str): The input text string to process.

    Returns:
        dict: A dictionary containing the cleaned text, character length,
              and estimated token count.
    """
    # 4. Error Handling for Edge Cases
    if not isinstance(text, str):
        raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")

    if not text.strip():
        return {
            "cleaned_text": "",
            "char_length": 0,
            "estimated_tokens": 0
        }

    # Normalize text: remove unwanted special characters and collapse extra whitespace
    cleaned = re.sub(r'[^\w\s.,!?-]', '', text)
    cleaned = re.sub(r'\s+', ' ', cleaned).strip()

    char_length = len(cleaned)
    # Standard rule of thumb: ~4 characters per token
    estimated_tokens = max(1, char_length // 4)

    return {
        "cleaned_text": cleaned,
        "char_length": char_length,
        "estimated_tokens": estimated_tokens
    }

# ==========================================
# 2 & 3. Test Cases & Verification
# ==========================================
if __name__ == "__main__":
    print("--- Test Case 1: Standard Text ---")
    sample_input = "   Hello! Welcome to the Digital Twin architecture course. Let's analyze token limits.   "
    result1 = estimate_and_clean_text(sample_input)
    print(f"Original Text: '{sample_input}'")
    print(f"Cleaned Text:  '{result1['cleaned_text']}'")
    print(f"Char Length:   {result1['char_length']}")
    print(f"Est. Tokens:   {result1['estimated_tokens']}")

    print("\n--- Test Case 2: Error Handling (Invalid Type) ---")
    try:
        estimate_and_clean_text(12345)
    except TypeError as e:
        print(f"Successfully caught expected TypeError: {e}")

    print("\n--- Test Case 3: Empty String Edge Case ---")
    result3 = estimate_and_clean_text("   ")
    print(f"Result for empty/whitespace input: {result3}")
```

```python
import re
import logging

logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger(__name__)

class DigitalTwinContextManager:
    """
    Intermediate implementation for managing digital twin context windows,
    combining text cleaning, token estimation, sliding window rules, and parameter validation.
    """

    def __init__(self, max_context_tokens: int = 256, overlap_tokens: int = 32):
        # 2 & 3. Parameter validation & helpful error messages
        if not isinstance(max_context_tokens, int) or max_context_tokens <= 0:
            raise ValueError("max_context_tokens must be a positive integer.")
        if not isinstance(overlap_tokens, int) or overlap_tokens < 0:
            raise ValueError("overlap_tokens must be a non-negative integer.")
        if overlap_tokens >= max_context_tokens:
            raise ValueError("overlap_tokens must be strictly less than max_context_tokens.")

        self.max_context_tokens = max_context_tokens
        self.overlap_tokens = overlap_tokens
        self.messages = []

    def estimate_tokens(self, text: str) -> int:
        """Estimates token count using the standard rule of thumb (~4 characters per token)."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")
        return max(1, len(text) // 4)

    def clean_text(self, text: str) -> str:
        """Normalizes and cleans input text by removing unwanted characters and excess whitespace."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")

        cleaned = re.sub(r'[^\w\s.,!?-]', '', text)
        return re.sub(r'\s+', ' ', cleaned).strip()

    def add_message(self, role: str, content: str):
        """Adds a conversational turn to the history with validated roles and token calculation."""
        if role not in ["system", "user", "assistant"]:
            raise ValueError("Role must be strictly 'system', 'user', or 'assistant'.")

        cleaned_content = self.clean_text(content)
        token_count = self.estimate_tokens(cleaned_content)

        self.messages.append({
            "role": role,
            "content": cleaned_content,
            "tokens": token_count
        })
        logger.debug(f"Added [{role}] message ({token_count} tokens).")

    def truncate_to_limit(self) -> list[dict]:
        """Assembles context within the token budget, preserving system instructions and recent turns."""
        if not self.messages:
            return []

        system_msgs = [m for m in self.messages if m["role"] == "system"]
        other_msgs = [m for m in self.messages if m["role"] != "system"]

        system_tokens = sum(m["tokens"] for m in system_msgs)
        budget = self.max_context_tokens - system_tokens
        if budget < 0:
            budget = 0

        selected_others = []
        current_tokens = 0

        for msg in reversed(other_msgs):
            if current_tokens + msg["tokens"] <= budget:
                selected_others.insert(0, msg)
                current_tokens += msg["tokens"]
            else:
                break

        return system_msgs + selected_others


# ==========================================
# 4. Testing with Various Input Types & Cases
# ==========================================
if __name__ == "__main__":
    print("--- Test Case 1: Standard Context Management ---")
    manager = DigitalTwinContextManager(max_context_tokens=120, overlap_tokens=20)

    manager.add_message("system", "You are an expert AI engineering collaborator.")
    manager.add_message("user", "Can you explain how embedding vectors represent semantics?")
    manager.add_message("assistant", "Embeddings map words into continuous vector space where semantic proximity aligns with directional angles.")
    manager.add_message("user", "How does cosine similarity measure this?")

    active_context = manager.truncate_to_limit()
    for msg in active_context:
        print(f"[{msg['role'].upper()}] ({msg['tokens']}t): {msg['content']}")

    print("\n--- Test Case 2: Parameter Validation Error (Invalid Overlap) ---")
    try:
        DigitalTwinContextManager(max_context_tokens=100, overlap_tokens=100)
    except ValueError as e:
        print(f"Caught expected ValueError: {e}")

    print("\n--- Test Case 3: Type Validation Error (Invalid Input Type) ---")
    try:
        manager.add_message("user", 99999)  # Passing integer instead of string
    except TypeError as e:
        print(f"Caught expected TypeError: {e}")
```

```python
import re
import logging
import matplotlib.pyplot as plt

# Configure structured logging for debugging and tracking execution
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - [%(filename)s:%(lineno)d] - %(message)s'
)
logger = logging.getLogger(__name__)

class ProductionSystem:
    """
    Production-ready Context Window & Token Budget Manager for Digital Twins.

    Features:
    - Text normalization and noise reduction via regex.
    - Subword token estimation heuristic (~4 characters per token).
    - Intelligent context window budgeting with system prompt preservation.
    - Priority-based message pruning to stay strictly within token limits.
    - Structured logging for operational visibility.
    - Automated result visualization (token budget allocation chart).
    - Comprehensive input validation and informative error handling.
    """

    def __init__(self, max_context_tokens: int = 512, overlap_tokens: int = 64):
        """
        Initializes the production system with token capacity and overlap bounds.

        Args:
            max_context_tokens (int): Maximum token capacity for the LLM context window.
            overlap_tokens (int): Token overlap for sliding chunk windows.
        """
        if not isinstance(max_context_tokens, int) or max_context_tokens <= 0:
            raise ValueError("max_context_tokens must be a positive integer.")
        if not isinstance(overlap_tokens, int) or overlap_tokens < 0:
            raise ValueError("overlap_tokens must be a non-negative integer.")
        if overlap_tokens >= max_context_tokens:
            raise ValueError("overlap_tokens must be strictly less than max_context_tokens.")

        self.max_context_tokens = max_context_tokens
        self.overlap_tokens = overlap_tokens
        self.messages = []
        logger.info(f"Initialized ProductionSystem (Max Context: {max_context_tokens}t, Overlap: {overlap_tokens}t)")

    def estimate_tokens(self, text: str) -> int:
        """Estimates token count using standard subword heuristic (~4 characters per token)."""
        if not isinstance(text, str):
            raise TypeError(f"Expected string input for token estimation, got {type(text).__name__}.")
        return max(1, len(text) // 4)

    def clean_text(self, text: str) -> str:
        """Normalizes input text by removing unauthorized characters and excess whitespace."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")
        cleaned = re.sub(r'[^\w\s.,!?-]', '', text)
        return re.sub(r'\s+', ' ', cleaned).strip()

    def add_interaction(self, role: str, content: str, priority: int = 1):
        """
        Adds a conversational turn or telemetry log to the digital twin history.

        Args:
            role (str): Role identifier ('system', 'user', or 'assistant').
            content (str): Raw text content.
            priority (int): Importance weight for pruning decisions (higher = more vital).
        """
        if role not in ["system", "user", "assistant"]:
            raise ValueError("Role must be strictly 'system', 'user', or 'assistant'.")
        if not isinstance(priority, int) or priority < 0:
            raise ValueError("Priority must be a non-negative integer.")

        cleaned_content = self.clean_text(content)
        token_count = self.estimate_tokens(cleaned_content)

        self.messages.append({
            "role": role,
            "content": cleaned_content,
            "tokens": token_count,
            "priority": priority
        })
        logger.debug(f"Added [{role}] interaction with {token_count} tokens (Priority: {priority}).")

    def process(self) -> list[dict]:
        """
        Assembles active context within the configured token budget, preserving
        system prompts and selecting recent messages by recency and priority.

        Returns:
            list[dict]: List of messages fitting within the active context window.
        """
        if not self.messages:
            return []

        system_msgs = [m for m in self.messages if m["role"] == "system"]
        other_msgs = [m for m in self.messages if m["role"] != "system"]

        system_tokens = sum(m["tokens"] for m in system_msgs)
        budget = self.max_context_tokens - system_tokens

        if budget < 0:
            logger.warning("System messages exceed total token budget limit!")
            budget = 0

        selected_others = []
        current_tokens = 0

        for msg in reversed(other_msgs):
            if current_tokens + msg["tokens"] <= budget:
                selected_others.insert(0, msg)
                current_tokens += msg["tokens"]
            else:
                logger.info(f"Pruned message ({msg['tokens']} tokens) due to context window limit.")

        final_context = system_msgs + selected_others
        total_used = sum(m["tokens"] for m in final_context)
        logger.info(f"Assembled context: {len(final_context)} messages consuming {total_used}/{self.max_context_tokens} tokens.")
        return final_context

    def visualize_results(self, save_path: str = "production_context_distribution.png") -> str:
        """
        Generates and saves a horizontal bar chart visualizing token allocation across the context window.

        Args:
            save_path (str): File path for saving the generated chart image.

        Returns:
            str: Path to the saved visualization file.
        """
        context = self.process()
        if not context:
            logger.warning("No context available to visualize.")
            return None

        labels = [f"{m['role'].upper()} ({m['tokens']}t)" for m in context]
        tokens = [m["tokens"] for m in context]
        colors = ['#10b981' if m['role']=='system' else '#3b82f6' if m['role']=='user' else '#8b5cf6' for m in context]

        plt.figure(figsize=(10, 5))
        plt.barh(labels, tokens, color=colors, edgecolor='#1f2937', height=0.6, linewidth=1.2)
        plt.axvline(x=self.max_context_tokens, color='red', linestyle='--', linewidth=2, label=f'Max Limit ({self.max_context_tokens}t)')
        plt.xlabel("Estimated Token Count", fontsize=12, fontweight='bold')
        plt.title("Production Digital Twin Context Window Allocation", fontsize=14, fontweight='bold', pad=15)
        plt.xlim(0, self.max_context_tokens * 1.2)
        plt.grid(axis='x', linestyle='--', alpha=0.6)
        plt.legend(loc='lower right', frameon=True)
        plt.tight_layout()

        plt.savefig(save_path, dpi=300)
        plt.close()
        logger.info(f"Context budget visualization successfully saved to '{save_path}'.")
        return save_path


# ==========================================
# Comprehensive Test Suite & Verification
# ==========================================
if __name__ == "__main__":
    print("=== Test Case 1: Initializing ProductionSystem ===")
    system = ProductionSystem(max_context_tokens=220, overlap_tokens=32)

    print("\n=== Test Case 2: Adding Interactions ===")
    system.add_interaction("system", "You are an expert mechanical engineering AI collaborator and youth leadership mentor.", priority=10)
    system.add_interaction("user", "Can you help me structure my final-year project analysis on mechanical properties of recycled tinplate?", priority=5)
    system.add_interaction("assistant", "Certainly! We can evaluate alkaline leaching, microstructural testing, and stress-strain curves compared to low-carbon steel.", priority=5)
    system.add_interaction("user", "What are the key optimization steps for thermodynamic turbomachinery cycles?", priority=3)

    active_context = system.process()
    for msg in active_context:
        print(f"[{msg['role'].upper()}] ({msg['tokens']}t): {msg['content']}")

    print("\n=== Test Case 3: Generating Visualization Chart ===")
    chart_file = system.visualize_results("exercise3_visualization.png")
    print(f"Visualization chart saved to: {chart_file}")

    print("\n=== Test Case 4: Error Handling (Invalid Parameter) ===")
    try:
        ProductionSystem(max_context_tokens=100, overlap_tokens=150)
    except ValueError as e:
        print(f"Caught expected ValueError: {e}")
```

```python
import re
import logging
import numpy as np
import matplotlib.pyplot as plt

# Configure structured logging for debugging and operational visibility
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - [%(filename)s:%(lineno)d] - %(message)s'
)
logger = logging.getLogger(__name__)

class DigitalTwinRAGSystem:
    """
    Production-Grade Retrieval-Augmented Generation (RAG) System for Digital Twins.

    Features:
    - Text normalization and noise reduction via regex.
    - Sliding-window smart chunking with configurable max size and overlap.
    - Source attribution metadata enrichment for precise traceability.
    - Vector embedding generation and semantic retrieval via Cosine Similarity.
    - Intelligent context window token budgeting.
    - Structured logging, comprehensive error handling, and visual analytics.
    """

    def __init__(self, max_context_tokens: int = 512, max_chunk_size: int = 150, overlap: int = 30):
        if not isinstance(max_context_tokens, int) or max_context_tokens <= 0:
            raise ValueError("max_context_tokens must be a positive integer.")
        if not isinstance(max_chunk_size, int) or max_chunk_size <= 0:
            raise ValueError("max_chunk_size must be a positive integer.")
        if not isinstance(overlap, int) or overlap < 0:
            raise ValueError("overlap must be a non-negative integer.")
        if overlap >= max_chunk_size:
            raise ValueError("overlap must be strictly less than max_chunk_size.")

        self.max_context_tokens = max_context_tokens
        self.max_chunk_size = max_chunk_size
        self.overlap = overlap
        self.knowledge_base = []
        logger.info(f"Initialized DigitalTwinRAGSystem (Max Context: {max_context_tokens}t, Chunk Size: {max_chunk_size}t)")

    def clean_text(self, text: str) -> str:
        """Normalizes input text by removing unwanted characters and excess whitespace."""
        if not isinstance(text, str):
            raise TypeError(f"Expected input type 'str', got '{type(text).__name__}' instead.")
        cleaned = re.sub(r'[^\w\s.,!?-]', '', text)
        return re.sub(r'\s+', ' ', cleaned).strip()

    def estimate_tokens(self, text: str) -> int:
        """Estimates token count using the standard subword heuristic (~4 characters per token)."""
        if not isinstance(text, str):
            raise TypeError("Text must be a string.")
        return max(1, len(text) // 4)

    def add_document(self, text: str, source: str):
        """
        Ingests a document, cleans it, splits it into overlapping chunks,
        generates embeddings, and stores them in the knowledge base.
        """
        cleaned_text = self.clean_text(text)
        if not cleaned_text:
            logger.warning(f"Document from source '{source}' resulted in empty text after cleaning.")
            return

        start = 0
        text_length = len(cleaned_text)
        chunk_id = 1

        while start < text_length:
            end = min(start + self.max_chunk_size, text_length)
            chunk_text = cleaned_text[start:end]

            # Generate mock normalized embedding vector (replace with API embedding in production)
            np.random.seed(hash(chunk_text) % 2**32)
            vector = np.random.randn(3)
            vector = vector / np.linalg.norm(vector)

            chunk_record = {
                "id": f"{source}_chunk_{chunk_id}",
                "text": chunk_text,
                "source": source,
                "length": len(chunk_text),
                "tokens": self.estimate_tokens(chunk_text),
                "embedding": vector
            }
            self.knowledge_base.append(chunk_record)

            chunk_id += 1
            start += self.max_chunk_size - self.overlap

        logger.info(f"Successfully ingested and chunked document from source '{source}' into {chunk_id-1} chunks.")

    def retrieve(self, query: str, top_k: int = 3) -> list:
        """Retrieves the top-k most relevant chunks using Cosine Similarity against query vector."""
        if not self.knowledge_base:
            logger.warning("Knowledge base is empty. No chunks to retrieve.")
            return []

        cleaned_query = self.clean_text(query)
        np.random.seed(hash(cleaned_query) % 2**32)
        query_vector = np.random.randn(3)
        query_vector = query_vector / np.linalg.norm(query_vector)

        scored_chunks = []
        for chunk in self.knowledge_base:
            # Cosine similarity for normalized vectors is their dot product
            similarity = float(np.dot(query_vector, chunk["embedding"]))
            scored_chunks.append((similarity, chunk))

        # Sort descending by similarity score
        scored_chunks.sort(key=lambda x: x[0], reverse=True)

        top_results = []
        for sim, chunk in scored_chunks[:top_k]:
            res = chunk.copy()
            res["similarity_score"] = round(sim, 4)
            top_results.append(res)

        logger.info(f"Retrieved {len(top_results)} chunks for query: '{query}'")
        return top_results

    def assemble_rag_context(self, query: str, chat_history: list = None) -> dict:
        """Assembles a complete RAG context package within the active token budget."""
        chat_history = chat_history or []
        history_tokens = sum(m.get("tokens", self.estimate_tokens(m.get("content", ""))) for m in chat_history)

        remaining_budget = self.max_context_tokens - history_tokens
        if remaining_budget < 50:
            logger.warning("Context budget highly constrained. Reducing retrieved chunks to 1.")
            top_chunks = self.retrieve(query, top_k=1)
        else:
            top_chunks = self.retrieve(query, top_k=3)

        selected_chunks = []
        chunk_tokens_total = 0

        for chunk in top_chunks:
            if chunk_tokens_total + chunk["tokens"] <= remaining_budget:
                selected_chunks.append(chunk)
                chunk_tokens_total += chunk["tokens"]

        return {
            "query": query,
            "chat_history": chat_history,
            "retrieved_chunks": selected_chunks,
            "total_tokens_used": history_tokens + chunk_tokens_total,
            "max_budget": self.max_context_tokens
        }

    def visualize_retrieval_performance(self, query: str, save_path: str = "rag_retrieval_analysis.png") -> str:
        """Generates and saves a horizontal bar chart visualizing retrieval cosine similarities."""
        results = self.retrieve(query, top_k=5)
        if not results:
            logger.warning("No results available to visualize.")
            return None

        labels = [c["id"] for c in results]
        scores = [c["similarity_score"] for c in results]

        plt.figure(figsize=(9, 5))
        plt.barh(labels, scores, color='#2563eb', edgecolor='#1e3a8a', height=0.5, linewidth=1.2)
        plt.axvline(x=0.0, color='grey', linestyle='--', linewidth=1)
        plt.xlabel("Cosine Similarity Score", fontsize=12, fontweight='bold')
        plt.title(f"RAG Retrieval Similarity for Query: '{query}'", fontsize=14, fontweight='bold', pad=15)
        plt.xlim(min(0, min(scores)-0.1), 1.05)
        plt.grid(axis='x', linestyle='--', alpha=0.6)
        plt.tight_layout()

        plt.savefig(save_path, dpi=300)
        plt.close()
        logger.info(f"Retrieval analysis chart successfully saved to '{save_path}'.")
        return save_path


# ==========================================
# Comprehensive Test Suite & Verification
# ==========================================
if __name__ == "__main__":
    print("=== Test Case 1: Initializing RAG System ===")
    rag_system = DigitalTwinRAGSystem(max_context_tokens=350, max_chunk_size=100, overlap=20)

    print("\n=== Test Case 2: Ingesting Documents ===")
    rag_system.add_document(
        "Digital twin architectures require seamless integration of streaming telemetry, historical activity sequences, and context-aware RAG pipelines. By chunking structured documents with controlled overlap and rich metadata attribution, LLMs maintain high precision across complex reasoning tasks.",
        source="Digital_Twin_Architecture_v2.pdf"
    )
    rag_system.add_document(
        "Mechanical properties of recycled tinplate cans evaluated via alkaline leaching and microstructural testing at UMaT Minerals Laboratory demonstrate excellent structural viability compared to low-carbon steel.",
        source="Tinplate_Thesis_Report.pdf"
    )

    print("\n=== Test Case 3: Assembling RAG Context Package ===")
    sample_query = "What are the mechanical properties of recycled tinplate?"
    context_pkg = rag_system.assemble_rag_context(sample_query)

    print(f"Query: {context_pkg['query']}")
    print(f"Total Tokens Used: {context_pkg['total_tokens_used']}/{context_pkg['max_budget']}")
    for i, chunk in enumerate(context_pkg['retrieved_chunks'], 1):
        print(f"\nRetrieved Chunk {i} [{chunk['source']} - Score: {chunk['similarity_score']}]:")
        print(f"  {chunk['text']}")

    print("\n=== Test Case 4: Generating Visualization Analysis ===")
    chart_file = rag_system.visualize_retrieval_performance(sample_query, "rag_retrieval_analysis.png")
    print(f"Retrieval chart successfully saved to: {chart_file}")
```
